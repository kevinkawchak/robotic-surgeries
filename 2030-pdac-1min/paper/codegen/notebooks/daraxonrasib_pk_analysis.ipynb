# Daraxonrasib Perioperative PK Analysis (v0.6.0)

```python
import math
import matplotlib.pyplot as plt

half_life_hours = 36.0
induction = 35.0
hours = list(range(0, 168))
serum = [induction * math.exp(-math.log(2)/half_life_hours * h) for h in hours]
plt.plot(hours, serum)
plt.axhline(0.5, color='red', linestyle='--', label='active threshold 0.5 ng/mL')
plt.xlabel('hours since last dose')
plt.ylabel('serum (ng/mL)')
plt.title('Daraxonrasib washout (one compartment, t1/2 = 36 h)')
plt.legend()
plt.tight_layout()
plt.savefig('../outputs/daraxonrasib/washout_curve.png')
plt.show()
```
