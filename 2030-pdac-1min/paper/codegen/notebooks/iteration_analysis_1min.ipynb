# 32 Iteration Cross Arm Analysis (v0.6.0)

This notebook analyzes the 32 iteration sweep outputs at `../data/iterations/`.

```python
import json
import pandas as pd

records = []
with open('../data/iterations/index.jsonl') as f:
    for line in f:
        records.append(json.loads(line))
df = pd.DataFrame(records)
print('iterations:', len(df))
print('mean composite:', df['composite_score'].mean().round(3))
print('95 ci half width:', (df['composite_score'].std() / (len(df) ** 0.5) * 1.96).round(3))
df.describe()
```

```python
print('PJ grade A rate:', (df['realized_pj_grade'] == 'A').mean().round(3))
print('PJ grade B rate:', (df['realized_pj_grade'] == 'B').mean().round(3))
print('PJ grade C rate:', (df['realized_pj_grade'] == 'C').mean().round(3))
print('HJ leak absent rate:', (df['realized_hj_leak'] == 'absent').mean().round(3))
print('GJ patent rate:', (df['realized_gj_patency'] == 'patent').mean().round(3))
```
